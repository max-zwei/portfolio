```python
## Add mean+2std (as of 2001)
# Create a function which takes an 2001 activity code and calculates the mean_2std based for it. 
# Join the 2001 mean_2std for the remaining years

mean_2std_2001 = (
    df_YBY_AG1529_described
    .filter(pl.col("year") == 2001)
    .select(
        "activity_code",
        (pl.col("mean") + 2 * pl.col("std")).alias("mean+2std (2001)")
    )
)

df_YBY_AG1529_described = df_YBY_AG1529_described.join(
    mean_2std_2001,
    on="activity_code",
    how="left",
)

## Add mean+2std for the row to compare to mean+2std
```

```python
## Add trends to compare year-over-year and a trend for 1991-2022 for each activity code for "df_Year-by-Year_AG1529_overview"

df_YBY_AG1529_described = df_YBY_AG1529_described.with_columns(
    (pl.col("2001") - pl.col("1991")).alias("1991 -> 2001")
)

df_YBY_AG1529_described = df_YBY_AG1529_described.with_columns(
    (pl.col("2012") - pl.col("2001")).alias("2001 -> 2012")
)

df_YBY_AG1529_described = df_YBY_AG1529_described.with_columns(
    (pl.col("2022") - pl.col("2012")).alias("2012 -> 2022")
)

df_YBY_AG1529_described = df_YBY_AG1529_described.with_columns(
    (pl.col("2022") - pl.col("1991")).alias("1991 -> 2022")
)
```

```python
## Create average and max incline/decline year over year

columns = ["1991 -> 2001", "2001 -> 2012", "2012 -> 2022"]
abs_values = pl.col(columns).replace(0, None).abs()

df_YBY_AG1529_described = df_YBY_AG1529_described.with_columns(
    pl.mean_horizontal(
        abs_values
    ).alias("year-over-year (mean)")
)

df_YBY_AG1529_described = df_YBY_AG1529_described.with_columns(
    pl.max_horizontal(
        abs_values
    )
    .alias("year-over-year (max)")
)
```
